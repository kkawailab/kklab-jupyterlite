## 1. pandas とは何か

**pandas** は、Python でデータ分析を行うためのライブラリです。

- **表形式のデータ**（Excel のスプレッドシートのようなもの）を簡単に扱える

### pandas の 2 つの主要なデータ構造

| **Series** | 1次元のラベル付き配列 | Excel の 1 列分 |

## 2. Series の基本

**Series** は、インデックス（ラベル）付きの 1 次元配列です。

### 2.1 Series の作成

````python
# リストから Series を作成
s1 = pd.Series([10, 20, 30, 40, 50])
print("リストから作成した Series:")
print(s1)
````

````python
# インデックスを指定して作成
s2 = pd.Series([85, 92, 78, 95], index=["数学", "英語", "国語", "理科"])
print("インデックス付きの Series:")
print(s2)
````

````python
# 辞書から Series を作成
prices = {"りんご": 150, "バナナ": 100, "みかん": 80, "ぶどう": 300}
s3 = pd.Series(prices)
print("辞書から作成した Series:")
print(s3)
````

### 2.2 Series の基本属性

````python
print(f"値（values）: {s2.values}")
print(f"インデックス（index）: {list(s2.index)}")
print(f"データ型（dtype）: {s2.dtype}")
print(f"要素数（size）: {s2.size}")
````

### 2.3 Series の要素へのアクセス

````python
# ラベルでアクセス
print(f"数学の点数: {s2['数学']}")

# 位置（番号）でアクセス
print(f"最初の要素: {s2.iloc[0]}")

# スライス
print(f"\n最初の2つ:")
print(s2.iloc[:2])
````

### 2.4 Series の演算

````python
# 全要素に対する演算
print("元の点数:")
print(s2)
print("\n全員に +5 点:")
print(s2 + 5)

print("\n点数を 100 で割って割合に:")
print(s2 / 100)
````

## 5. データの選択とスライス

DataFrame から特定の行や列を取り出す方法を学びます。

````python
# サンプルデータの確認
df
````

### 5.1 列の選択

````python
# 1 列を選択（Series が返る）
df["名前"]
````

````python
# 複数列を選択（DataFrame が返る）
df[["名前", "年齢", "給与"]]
````

### 5.2 loc と iloc の違い

| メソッド | 指定方法 | 説明 |
|---------|---------|------|
| `loc` | ラベル | 行名・列名で指定 |
| `iloc` | 位置（番号）| 行番号・列番号で指定（0 から始まる）|

````python
# loc: ラベルで選択
print("loc[0, '名前']:")
print(df.loc[0, "名前"])

print("\nloc[1:3, ['名前', '年齢']]:")
df.loc[1:3, ["名前", "年齢"]]  # 1〜3 行目（3 を含む）
````

````python
# iloc: 位置（番号）で選択
print("iloc[0, 0]:")
print(df.iloc[0, 0])  # 0 行目、0 列目

print("\niloc[1:3, 0:2]:")
df.iloc[1:3, 0:2]  # 1〜2 行目（3 は含まない）、0〜1 列目
````

### 5.3 loc と iloc の重要な違い

- `loc[1:3]` → 1, 2, 3 行目（**終端を含む**）
- `iloc[1:3]` → 1, 2 行目（**終端を含まない**、Python のスライスと同じ）

````python
print("loc[0:2] - ラベル 0, 1, 2 の行（3 行）:")
print(df.loc[0:2, "名前"].tolist())

print("\niloc[0:2] - 位置 0, 1 の行（2 行）:")
print(df.iloc[0:2, 0].tolist())
````

## 6. 条件によるフィルタリング

条件を指定して、特定の行だけを抽出する方法を学びます。

### 6.1 単一条件でのフィルタリング

````python
# 年齢が 25 歳以上の人
df[df["年齢"] >= 25]
````

````python
# 条件式の中身を確認（True/False の Series）
print("df['年齢'] >= 25 の結果:")
print(df["年齢"] >= 25)
````

````python
# 文字列の条件
df[df["部署"] == "営業"]
````

### 6.2 複数条件でのフィルタリング

複数条件を組み合わせるときは:
- AND（かつ）: `&`
- OR（または）: `|`
- NOT（否定）: `~`

**重要**: 各条件を `()` で囲むこと！

````python
# 営業部 かつ 年齢が 25 歳以上
df[(df["部署"] == "営業") & (df["年齢"] >= 25)]
````

````python
# 営業部 または 開発部
df[(df["部署"] == "営業") | (df["部署"] == "開発")]
````

````python
# isin を使うと複数の値を簡潔に指定できる
df[df["部署"].isin(["営業", "開発"])]
````

### 6.3 query メソッド

文字列で条件を書ける便利なメソッドです。

````python
# query を使った書き方
df.query("年齢 >= 25 and 部署 == '営業'")
````

## 7. 基本的な統計量の計算

pandas では、基本的な統計量を簡単に計算できます。

### 7.1 describe で概要を確認

````python
# 数値列の基本統計量
df.describe()
````

### 7.2 個別の統計量

````python
print(f"給与の平均: {df['給与'].mean():,.0f} 円")
print(f"給与の中央値: {df['給与'].median():,.0f} 円")
print(f"給与の標準偏差: {df['給与'].std():,.0f} 円")
print(f"給与の最大値: {df['給与'].max():,.0f} 円")
print(f"給与の最小値: {df['給与'].min():,.0f} 円")
print(f"給与の合計: {df['給与'].sum():,.0f} 円")
````

### 7.3 value_counts でカテゴリの頻度を確認

### 7.4 相関係数
