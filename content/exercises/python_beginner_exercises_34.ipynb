````python
# 問題 33 の解答
np.random.seed(42)
group_A = np.random.normal(loc=65, scale=10, size=30)
group_B = np.random.normal(loc=70, scale=10, size=30)

# 1. 平均と標準偏差
print("グループ A:")
print(f"  平均: {group_A.mean():.2f}, 標準偏差: {group_A.std():.2f}")
print("グループ B:")
print(f"  平均: {group_B.mean():.2f}, 標準偏差: {group_B.std():.2f}")

# 2. t 検定
t_stat, p_value = stats.ttest_ind(group_A, group_B)

# 3. 結果の表示
print(f"\nt 値: {t_stat:.4f}")
print(f"p 値: {p_value:.4f}")
if p_value < 0.05:
    print("→ 有意水準 5% で有意差あり")
else:
    print("→ 有意水準 5% で有意差なし")

# 4. 箱ひげ図
df_ttest = pd.DataFrame({
    "score": np.concatenate([group_A, group_B]),
    "group": ["A"] * 30 + ["B"] * 30
})

plt.figure(figsize=(6, 5))
sns.boxplot(data=df_ttest, x="group", y="score")
plt.xlabel("グループ")
plt.ylabel("スコア")
plt.title("2 群の分布比較")
plt.show()
````

````python
# 問題 34 の解答
from statsmodels.stats.anova import anova_lm

np.random.seed(42)
group_A = np.random.normal(loc=60, scale=8, size=25)
group_B = np.random.normal(loc=65, scale=8, size=25)
group_C = np.random.normal(loc=70, scale=8, size=25)

# 1. DataFrame の作成
score = np.concatenate([group_A, group_B, group_C])
group = ["A"] * 25 + ["B"] * 25 + ["C"] * 25
df_anova = pd.DataFrame({"score": score, "group": group})

# 2. ANOVA の実行
model_anova = smf.ols("score ~ C(group)", data=df_anova).fit()
anova_table = anova_lm(model_anova)

# 3. 結果の表示
print("=== ANOVA 結果 ===")
print(anova_table)
print()

f_value = anova_table.loc["C(group)", "F"]
p_value = anova_table.loc["C(group)", "PR(>F)"]
print(f"F 値: {f_value:.4f}")
print(f"p 値: {p_value:.4f}")
if p_value < 0.05:
    print("→ 有意水準 5% で群間に有意差あり")
else:
    print("→ 有意水準 5% で群間に有意差なし")

# 4. 箱ひげ図
plt.figure(figsize=(6, 5))
sns.boxplot(data=df_anova, x="group", y="score")
plt.xlabel("グループ")
plt.ylabel("スコア")
plt.title("3 群の分布比較（ANOVA）")
plt.show()
````
