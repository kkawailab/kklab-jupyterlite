## 問題 19: 二元配置分散分析

以下のデータを使って二元配置分散分析を行いなさい。

```python
np.random.seed(42)
n_per_group = 15
factor_a = np.repeat(['A1', 'A2'], n_per_group * 2)
factor_b = np.tile(np.repeat(['B1', 'B2'], n_per_group), 2)
y = (np.where(factor_a == 'A1', 0, 5) + 
     np.where(factor_b == 'B1', 0, 3) + 
     np.random.normal(0, 2, n_per_group * 4))
df = pd.DataFrame({'A': factor_a, 'B': factor_b, 'y': y})
```

1. y ~ C(A) + C(B) + C(A):C(B) のモデルを推定する（交互作用を含む）
2. ANOVA表を作成する
3. 各要因と交互作用の有意性を確認する

````python
# 問題 19 の解答欄
np.random.seed(42)
n_per_group = 15
factor_a = np.repeat(['A1', 'A2'], n_per_group * 2)
factor_b = np.tile(np.repeat(['B1', 'B2'], n_per_group), 2)
y = (np.where(factor_a == 'A1', 0, 5) + 
     np.where(factor_b == 'B1', 0, 3) + 
     np.random.normal(0, 2, n_per_group * 4))
df = pd.DataFrame({'A': factor_a, 'B': factor_b, 'y': y})
````
