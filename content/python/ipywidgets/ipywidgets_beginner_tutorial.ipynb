## 1. ipywidgets とは何か

**ipywidgets** は、Jupyter Notebook/JupyterLite 上でインタラクティブな UI（ユーザーインターフェース）を作成するためのライブラリです。

### ipywidgets の特徴
- **インタラクティブ性**: スライダー、ボタン、テキスト入力などを使って、リアルタイムにデータを操作できる
- **コード不要の操作**: パラメータの調整をコード編集なしで行える
- **データ可視化との連携**: matplotlib や他のライブラリと組み合わせて、動的なグラフを作成できる
- **教育・プレゼンに最適**: 複雑な概念を視覚的・体験的に理解させることができる

### 基本的な使い方
```python
import ipywidgets as widgets
from IPython.display import display

# ウィジェットを作成
slider = widgets.IntSlider(value=50, min=0, max=100)

# ウィジェットを表示
display(slider)
```

## 2. 基本的なウィジェットの種類

ipywidgets には様々な種類のウィジェットがあります。用途に応じて使い分けましょう。

### 2.1 数値入力ウィジェット

````python
# IntSlider: 整数のスライダー
int_slider = widgets.IntSlider(
    value=50,           # 初期値
    min=0,              # 最小値
    max=100,            # 最大値
    step=1,             # ステップ
    description='整数:',  # ラベル
    continuous_update=True  # ドラッグ中も更新
)
display(int_slider)
````

````python
# FloatSlider: 小数のスライダー
float_slider = widgets.FloatSlider(
    value=3.14,
    min=0.0,
    max=10.0,
    step=0.01,
    description='小数:',
    readout_format='.2f'  # 表示形式
)
display(float_slider)
````

````python
# IntText / FloatText: 数値入力ボックス
int_text = widgets.IntText(
    value=42,
    description='数値入力:'
)
display(int_text)
````

````python
# BoundedIntText: 範囲制限付き数値入力
bounded_text = widgets.BoundedIntText(
    value=50,
    min=0,
    max=100,
    description='制限付き:'
)
display(bounded_text)
````

````python
# IntRangeSlider: 範囲選択スライダー
range_slider = widgets.IntRangeSlider(
    value=[20, 80],     # 初期範囲
    min=0,
    max=100,
    step=1,
    description='範囲:'
)
display(range_slider)
````

## 4. イベント処理（observe, on_click）

ウィジェットの値が変わったときや、ボタンがクリックされたときに処理を実行できます。

### 4.1 observe: 値の変更を監視

````python
# observe を使った値の監視
slider_obs = widgets.IntSlider(value=50, min=0, max=100, description='値:')
output_obs = widgets.Output()

def on_value_change(change):
    with output_obs:
        clear_output(wait=True)
        print(f"古い値: {change['old']}")
        print(f"新しい値: {change['new']}")

slider_obs.observe(on_value_change, names='value')

display(slider_obs, output_obs)
````

````python
# ドロップダウンの変更を監視
color_dropdown = widgets.Dropdown(
    options=['赤', '青', '緑', '黄'],
    value='赤',
    description='色:'
)
color_output = widgets.Output()

def on_color_change(change):
    color_map = {'赤': 'red', '青': 'blue', '緑': 'green', '黄': 'yellow'}
    with color_output:
        clear_output(wait=True)
        color = color_map[change['new']]
        display(widgets.HTML(
            f"<div style='width:100px;height:50px;background:{color}'></div>"
        ))

color_dropdown.observe(on_color_change, names='value')

# 初期表示
on_color_change({'new': color_dropdown.value})

display(color_dropdown, color_output)
````

### 4.2 on_click: ボタンのクリック処理

````python
# ボタンクリックのイベント処理
click_button = widgets.Button(description='クリック!', button_style='primary')
click_output = widgets.Output()
click_count = [0]  # リストにして参照を保持

def on_button_click(b):
    click_count[0] += 1
    with click_output:
        clear_output(wait=True)
        print(f"クリック回数: {click_count[0]} 回")

click_button.on_click(on_button_click)

display(click_button, click_output)
````

````python
# テキスト入力とボタンの組み合わせ
name_input = widgets.Text(placeholder='名前を入力', description='名前:')
greet_button = widgets.Button(description='挨拶する', button_style='success')
greet_output = widgets.Output()

def on_greet_click(b):
    name = name_input.value
    with greet_output:
        clear_output(wait=True)
        if name:
            print(f"こんにちは、{name}さん!")
        else:
            print("名前を入力してください")

greet_button.on_click(on_greet_click)

display(name_input, greet_button, greet_output)
````

### 4.3 link / dlink: ウィジェット間の連携

````python
# 2つのウィジェットを双方向にリンク
slider_a = widgets.IntSlider(description='スライダーA:', value=50)
slider_b = widgets.IntSlider(description='スライダーB:', value=50)

# 双方向リンク（どちらを動かしても同期する）
link = widgets.link((slider_a, 'value'), (slider_b, 'value'))

display(slider_a, slider_b)
````

````python
# 単方向リンク（ソース → ターゲット）
source_slider = widgets.IntSlider(description='ソース:', value=50)
target_progress = widgets.IntProgress(description='進捗:', value=50, bar_style='info')

# 単方向リンク
dlink = widgets.dlink((source_slider, 'value'), (target_progress, 'value'))

display(source_slider, target_progress)
````

## 5. interact による簡単なインタラクティブ関数

`interact` を使うと、関数の引数を自動的にウィジェットに変換し、インタラクティブに操作できます。

````python
from ipywidgets import interact, interactive, fixed
````

### 5.1 基本的な interact

````python
# 引数の型から自動的にウィジェットを生成
def greet(name):
    print(f"こんにちは、{name}さん!")

interact(greet, name='世界');
````

````python
# 整数 → IntSlider
def square(x):
    print(f"{x} の 2 乗は {x ** 2} です")

interact(square, x=10);
````

````python
# タプルで範囲を指定 (min, max) または (min, max, step)
def show_value(x):
    print(f"値: {x}")

interact(show_value, x=(0, 100, 5));
````

````python
# 小数 → FloatSlider
def show_float(x):
    print(f"値: {x:.2f}")

interact(show_float, x=(0.0, 1.0, 0.01));
````

````python
# ブール値 → Checkbox
def show_bool(flag):
    print(f"フラグ: {'ON' if flag else 'OFF'}")

interact(show_bool, flag=True);
````

````python
# リスト → Dropdown
def show_choice(item):
    print(f"選択: {item}")

interact(show_choice, item=['りんご', 'バナナ', 'みかん']);
````

### 5.2 複数の引数

````python
# 複数の引数を持つ関数
def calculate(a, b, operation):
    if operation == '足し算':
        result = a + b
    elif operation == '引き算':
        result = a - b
    elif operation == '掛け算':
        result = a * b
    else:  # 割り算
        result = a / b if b != 0 else '定義なし'
    print(f"{a} {operation} {b} = {result}")

interact(calculate, 
         a=(0, 100, 1), 
         b=(1, 100, 1),
         operation=['足し算', '引き算', '掛け算', '割り算']);
````

### 5.3 デコレータとして使う

````python
# @ デコレータとして使用
@interact(x=(0, 10), y=(0, 10))
def add(x, y):
    print(f"{x} + {y} = {x + y}")
````

### 5.4 interactive で結果を取得

````python
# interactive は結果オブジェクトを返す
def power(base, exp):
    return base ** exp

w = interactive(power, base=(2, 10), exp=(1, 5))
display(w)
````

````python
# 結果にアクセス
print(f"現在の結果: {w.result}")
print(f"引数: {w.kwargs}")
````

## 7. 複数ウィジェットの連携

複数のウィジェットを組み合わせて、より複雑なインターフェースを作成します。

### 7.1 フォームの作成

````python
# 入力フォームの作成
form_name = widgets.Text(description='名前:', placeholder='山田太郎')
form_age = widgets.IntSlider(description='年齢:', min=0, max=120, value=20)
form_gender = widgets.RadioButtons(description='性別:', options=['男性', '女性', 'その他'])
form_hobbies = widgets.SelectMultiple(
    description='趣味:',
    options=['読書', 'スポーツ', '音楽', 'ゲーム', '料理', '旅行'],
    rows=4
)
form_submit = widgets.Button(description='送信', button_style='primary')
form_output = widgets.Output()

def on_submit(b):
    with form_output:
        clear_output(wait=True)
        print("=== 入力内容 ===")
        print(f"名前: {form_name.value}")
        print(f"年齢: {form_age.value}歳")
        print(f"性別: {form_gender.value}")
        print(f"趣味: {', '.join(form_hobbies.value)}")

form_submit.on_click(on_submit)

form = widgets.VBox([
    widgets.HTML('<h3>プロフィール入力フォーム</h3>'),
    form_name,
    form_age,
    form_gender,
    form_hobbies,
    form_submit,
    form_output
])

display(form)
````

### 7.2 条件付き表示

````python
# チェックボックスで表示/非表示を切り替え
show_options = widgets.Checkbox(description='詳細オプションを表示', value=False)
detail_options = widgets.VBox([
    widgets.IntSlider(description='設定1:'),
    widgets.IntSlider(description='設定2:'),
    widgets.IntSlider(description='設定3:')
])

# 初期状態は非表示
detail_options.layout.display = 'none'

def toggle_options(change):
    if change['new']:
        detail_options.layout.display = 'flex'
    else:
        detail_options.layout.display = 'none'

show_options.observe(toggle_options, names='value')

display(show_options, detail_options)
````

### 8.1 単位変換器

````python
# 単位変換器
conv_value = widgets.FloatText(value=1.0, description='値:')
conv_from = widgets.Dropdown(
    options=['メートル', 'キロメートル', 'マイル', 'フィート'],
    value='メートル',
    description='変換元:'
)
conv_to = widgets.Dropdown(
    options=['メートル', 'キロメートル', 'マイル', 'フィート'],
    value='キロメートル',
    description='変換先:'
)
conv_result = widgets.HTML(value='<p style="font-size:18px">結果: </p>')

# メートルへの変換係数
to_meter = {
    'メートル': 1,
    'キロメートル': 1000,
    'マイル': 1609.34,
    'フィート': 0.3048
}

def convert(change=None):
    value = conv_value.value
    from_unit = conv_from.value
    to_unit = conv_to.value
    
    # 一度メートルに変換してから目的の単位に変換
    meters = value * to_meter[from_unit]
    result = meters / to_meter[to_unit]
    
    conv_result.value = f'<p style="font-size:18px"><b>{value} {from_unit}</b> = <b>{result:.4f} {to_unit}</b></p>'

# 各ウィジェットの変更を監視
conv_value.observe(convert, names='value')
conv_from.observe(convert, names='value')
conv_to.observe(convert, names='value')

# 初期計算
convert()

converter_ui = widgets.VBox([
    widgets.HTML('<h3>単位変換器</h3>'),
    conv_value,
    conv_from,
    conv_to,
    conv_result
])

display(converter_ui)
````

### 8.2 簡易電卓

````python
# 簡易電卓
calc_display = widgets.Text(
    value='0',
    disabled=True,
    layout=widgets.Layout(width='220px')
)

current_value = [0]
pending_operation = [None]
new_number = [True]

def make_button(text, style=''):
    btn = widgets.Button(
        description=text,
        layout=widgets.Layout(width='50px', height='40px'),
        button_style=style
    )
    return btn

def on_digit(b):
    if new_number[0]:
        calc_display.value = b.description
        new_number[0] = False
    else:
        if calc_display.value == '0':
            calc_display.value = b.description
        else:
            calc_display.value += b.description

def on_operation(b):
    if pending_operation[0] is not None:
        on_equals(None)
    current_value[0] = float(calc_display.value)
    pending_operation[0] = b.description
    new_number[0] = True

def on_equals(b):
    if pending_operation[0] is None:
        return
    second = float(calc_display.value)
    op = pending_operation[0]
    if op == '+':
        result = current_value[0] + second
    elif op == '-':
        result = current_value[0] - second
    elif op == '×':
        result = current_value[0] * second
    elif op == '÷':
        result = current_value[0] / second if second != 0 else 'Error'
    
    if result != 'Error':
        calc_display.value = str(result) if result == int(result) else f'{result:.6g}'
    else:
        calc_display.value = 'Error'
    pending_operation[0] = None
    new_number[0] = True

def on_clear(b):
    calc_display.value = '0'
    current_value[0] = 0
    pending_operation[0] = None
    new_number[0] = True

# ボタン作成
digits = [make_button(str(i)) for i in range(10)]
for d in digits:
    d.on_click(on_digit)

ops = [make_button(op, 'info') for op in ['+', '-', '×', '÷']]
for op in ops:
    op.on_click(on_operation)

equals_btn = make_button('=', 'success')
equals_btn.on_click(on_equals)

clear_btn = make_button('C', 'danger')
clear_btn.on_click(on_clear)

# レイアウト
row1 = widgets.HBox([digits[7], digits[8], digits[9], ops[3]])  # 7 8 9 ÷
row2 = widgets.HBox([digits[4], digits[5], digits[6], ops[2]])  # 4 5 6 ×
row3 = widgets.HBox([digits[1], digits[2], digits[3], ops[1]])  # 1 2 3 -
row4 = widgets.HBox([clear_btn, digits[0], equals_btn, ops[0]]) # C 0 = +

calculator = widgets.VBox([
    widgets.HTML('<h3>簡易電卓</h3>'),
    calc_display,
    row1, row2, row3, row4
])

display(calculator)
````

### 8.3 ToDo リスト

````python
# ToDo リスト
todo_input = widgets.Text(placeholder='新しいタスクを入力...', layout=widgets.Layout(width='250px'))
todo_add_btn = widgets.Button(description='追加', button_style='primary')
todo_list_container = widgets.VBox()

todos = []

def update_list():
    items = []
    for i, (task, done) in enumerate(todos):
        checkbox = widgets.Checkbox(value=done, indent=False, layout=widgets.Layout(width='30px'))
        label_style = 'text-decoration: line-through; color: gray;' if done else ''
        label = widgets.HTML(f'<span style="{label_style}">{task}</span>')
        delete_btn = widgets.Button(description='×', button_style='danger', layout=widgets.Layout(width='30px'))
        
        def make_toggle(idx):
            def toggle(change):
                todos[idx] = (todos[idx][0], change['new'])
                update_list()
            return toggle
        
        def make_delete(idx):
            def delete(b):
                todos.pop(idx)
                update_list()
            return delete
        
        checkbox.observe(make_toggle(i), names='value')
        delete_btn.on_click(make_delete(i))
        
        items.append(widgets.HBox([checkbox, label, delete_btn]))
    
    todo_list_container.children = items

def add_todo(b):
    task = todo_input.value.strip()
    if task:
        todos.append((task, False))
        todo_input.value = ''
        update_list()

todo_add_btn.on_click(add_todo)

# Enter キーでも追加
def on_enter(change):
    if change['new']:
        add_todo(None)

todo_ui = widgets.VBox([
    widgets.HTML('<h3>ToDo リスト</h3>'),
    widgets.HBox([todo_input, todo_add_btn]),
    todo_list_container
])

display(todo_ui)
````

### 8.4 インタラクティブなグラフ（Matplotlib 連携）

````python
# matplotlib のインストール（必要な場合）
try:
    import piplite
    await piplite.install(["matplotlib", "numpy"])
except ImportError:
    pass

import numpy as np
import matplotlib.pyplot as plt
````

````python
# インタラクティブなサイン波プロット
freq_slider = widgets.FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='周波数:')
amp_slider = widgets.FloatSlider(value=1.0, min=0.1, max=2.0, step=0.1, description='振幅:')
phase_slider = widgets.FloatSlider(value=0.0, min=0.0, max=2*np.pi, step=0.1, description='位相:')
plot_output = widgets.Output()

def update_plot(change=None):
    with plot_output:
        clear_output(wait=True)
        
        freq = freq_slider.value
        amp = amp_slider.value
        phase = phase_slider.value
        
        x = np.linspace(0, 2*np.pi, 500)
        y = amp * np.sin(freq * x + phase)
        
        fig, ax = plt.subplots(figsize=(8, 4))
        ax.plot(x, y, 'b-', linewidth=2)
        ax.set_xlim(0, 2*np.pi)
        ax.set_ylim(-2.5, 2.5)
        ax.set_xlabel('x')
        ax.set_ylabel('y')
        ax.set_title(f'y = {amp:.1f} × sin({freq:.1f}x + {phase:.1f})')
        ax.grid(True, alpha=0.3)
        ax.axhline(y=0, color='k', linewidth=0.5)
        plt.tight_layout()
        plt.show()

# 各スライダーの変更を監視
freq_slider.observe(update_plot, names='value')
amp_slider.observe(update_plot, names='value')
phase_slider.observe(update_plot, names='value')

# 初期プロット
update_plot()

plot_ui = widgets.VBox([
    widgets.HTML('<h3>インタラクティブなサイン波</h3>'),
    freq_slider,
    amp_slider,
    phase_slider,
    plot_output
])

display(plot_ui)
````
